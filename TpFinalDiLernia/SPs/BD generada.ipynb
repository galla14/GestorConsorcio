# [dbo].[sp_BajaLogicaUsuario]

```sql
/****** Object:  StoredProcedure [dbo].[sp_BajaLogicaUsuario]    Script Date: 28/10/2026 01:08:40 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
CREATE PROCEDURE [dbo].[sp_BajaLogicaUsuario]
	@NombreUsuario NVARCHAR(12)
AS
BEGIN
	
	IF NOT EXISTS(SELECT * FROM dbo.Usuario WHERE NombreUsuario = @NombreUsuario)
		BEGIN
			return -1;
		END

	SET NOCOUNT ON;
	UPDATE dbo.Usuario
	SET Activo = 0
	WHERE NombreUsuario = @NombreUsuario

	return 1;
END
GO
```

# [dbo].[sp_ConsultarExpensas]

```sql
/****** Object:  StoredProcedure [dbo].[sp_ConsultarExpensas]    Script Date: 28/10/2026 01:08:40 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
CREATE PROCEDURE [dbo].[sp_ConsultarExpensas]
AS
BEGIN
	SET NOCOUNT ON;

	SELECT  e.IdExpensas,
	        e.FechaEmision,
	        e.FechaVto,
	        e.MontoOrdinario,
	        e.MontoExtraOrd,
	        e.RecargoXMora,
	        e.Estado,
	        u.IdUnidad,
	        u.Piso,
	        u.Departamento,
	        u.MedidaDpto,
	        u.AptoAlquiler,
	        u.Alquilado,
	        p.IdPropietario,
	        p.Nombre,
	        p.Apellido,
	        p.Dni,
	        p.NroCelular
	FROM    dbo.Expensas e
	        INNER JOIN dbo.Unidad u      ON u.IdUnidad = e.IdUnidad
	        INNER JOIN dbo.Propietario p ON p.IdPropietario = u.IdPropietario;
END
GO
```

# [dbo].[sp_ConsultarUsuarioIniciarSesion]

```sql
/****** Object:  StoredProcedure [dbo].[sp_ConsultarUsuarioIniciarSesion]    Script Date: 28/10/2026 01:08:40 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
CREATE   PROCEDURE [dbo].[sp_ConsultarUsuarioIniciarSesion]
    @NombreUsuario NVARCHAR(50),
    @Password      NVARCHAR(50)
AS
BEGIN
    SET NOCOUNT ON;

    DECLARE @PassReal   NVARCHAR(12),
            @Intentos   INT,
            @Activo     BIT,
            @Bloqueado  BIT,
            @RolUsuario NVARCHAR(50);

    BEGIN TRY
        SELECT @PassReal   = Password,
               @Intentos   = IntentosFallidos,
               @Activo     = Activo,
               @Bloqueado  = Bloqueado,
               @RolUsuario = Rol
        FROM   dbo.Usuario
        WHERE  NombreUsuario = @NombreUsuario;

        
        IF @PassReal IS NULL OR @Activo = 0
            RETURN -1;

        
        IF @Bloqueado = 1
            RETURN -2;

        
        IF @PassReal <> @Password COLLATE Latin1_General_CS_AS
        BEGIN
            
            IF @RolUsuario = 'superadmin'
                RETURN -1;

            SET @Intentos = @Intentos + 1;

            
            IF @Intentos >= 3
            BEGIN
                UPDATE dbo.Usuario
                SET    IntentosFallidos = @Intentos, Bloqueado = 1
                WHERE  NombreUsuario = @NombreUsuario;

                RETURN -3;
            END

            
            UPDATE dbo.Usuario
            SET    IntentosFallidos = @Intentos
            WHERE  NombreUsuario = @NombreUsuario;

            RETURN -1;
        END

        
        UPDATE dbo.Usuario
        SET    IntentosFallidos = 0
        WHERE  NombreUsuario = @NombreUsuario;

        IF @RolUsuario = 'superadmin'
            RETURN 1;

        RETURN 2;
    END TRY
    BEGIN CATCH
        THROW;
    END CATCH
END
GO
```
